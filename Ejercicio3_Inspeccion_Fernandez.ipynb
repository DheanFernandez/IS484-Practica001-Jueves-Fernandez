# Inspección visual de un producto

Sistema inteligente encargado de analizar imágenes de productos
en una línea de producción y detectar posibles defectos mediante
el conteo de píxeles defectuosos.

## Ficha PEAS

| Elemento | Descripción |
|---|---|
| Percepción (S) | Imagen de 5x5 píxeles representada como matriz NumPy donde 1 representa un defecto |
| Acciones (A) | Aprobar producto, enviar a revisión manual o rechazar producto |
| Entorno (E) | Línea de producción industrial, productos fabricados y cámara de inspección |
| Objetivo | Detectar productos defectuosos antes de llegar al cliente |
| Medida de desempeño (P) | Precisión en la detección de defectos y reducción de productos defectuosos entregados |

## Justificación de las reglas

Las reglas permiten clasificar los productos según la cantidad de
defectos encontrados. Un producto sin defectos puede continuar su
proceso automáticamente, mientras que pequeñas fallas requieren
una revisión humana y una cantidad elevada de defectos indica
un posible producto no conforme.
