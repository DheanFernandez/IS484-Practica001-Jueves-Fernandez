# Escalamiento de tickets de soporte técnico

Sistema inteligente encargado de asignar el nivel de atención
adecuado para solicitudes de soporte considerando urgencia,
tiempo de espera y tipo de cliente.

## Ficha PEAS

| Elemento | Descripción |
|---|---|
| Percepción (S) | Tiempo de espera del ticket, nivel de urgencia reportado y tipo de cliente |
| Acciones (A) | Escalar ticket a nivel 1, nivel 2 o nivel 3 de soporte |
| Entorno (E) | Mesa de ayuda, clientes, técnicos de soporte y solicitudes registradas |
| Objetivo | Resolver incidencias priorizando los casos más críticos |
| Medida de desempeño (P) | Tiempo de respuesta, correcta clasificación de tickets y satisfacción del cliente |

## Justificación de las reglas

Las reglas buscan priorizar los problemas más críticos y reducir tiempos
de atención. Un ticket con alta urgencia debe recibir atención inmediata,
mientras que el tiempo de espera y la categoría del cliente ayudan a
determinar la prioridad cuando la urgencia no es suficiente.

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):

    # Convertimos la urgencia a minúsculas
    nivel_urgencia = nivel_urgencia.lower()

    # Caso crítico: urgencia alta
    if nivel_urgencia == "alta":

        return (
            "escalar a nivel 3",
            "urgencia alta requiere atención inmediata"
        )


    # Cliente premium con espera prolongada
    elif cliente_premium and tiempo_espera_minutos > 30:

        return (
            "escalar a nivel 2",
            "cliente premium con tiempo de espera elevado"
        )


    # Urgencia media
    elif nivel_urgencia == "media":

        if tiempo_espera_minutos > 60:

            return (
                "escalar a nivel 2",
                "urgencia media con espera prolongada"
            )

        else:

            return (
                "escalar a nivel 1",
                "urgencia media dentro del tiempo permitido"
            )


    # Urgencia baja
    else:

        if tiempo_espera_minutos > 120:

            return (
                "escalar a nivel 2",
                "ticket de baja urgencia pero con espera excesiva"
            )

        else:

            return (
                "escalar a nivel 1",
                "ticket de baja urgencia"
            )
```

```python
print(agente_soporte(10, "alta", False))

print(agente_soporte(40, "baja", True))

print(agente_soporte(70, "media", False))
```
