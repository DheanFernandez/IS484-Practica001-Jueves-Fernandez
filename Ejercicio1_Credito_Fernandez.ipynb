# Evaluación de solicitud de crédito bancario

Sistema inteligente encargado de evaluar solicitudes de crédito considerando
el ingreso mensual del cliente, monto solicitado y existencia de historial moroso.

## Ficha PEAS

| Elemento | Descripción |
|---|---|
| Percepción (S) | Ingreso mensual del cliente, monto solicitado e historial moroso |
| Acciones (A) | Aprobar crédito, aprobar con condiciones o rechazar solicitud |
| Entorno (E) | Clientes bancarios, solicitudes de crédito y políticas del banco |
| Objetivo | Evaluar el riesgo crediticio y tomar una decisión adecuada |
| Medida de desempeño (P) | Precisión en las decisiones, reducción de riesgos y correcta evaluación del cliente |

## Justificación de las reglas

Las reglas consideran la capacidad económica del cliente mediante la relación entre
el monto solicitado y su ingreso mensual. Además, se toma en cuenta el historial
moroso porque representa un factor importante de riesgo para la institución bancaria.
El objetivo es aprobar créditos viables y reducir posibles incumplimientos.

```python
def agente_credito(ingreso_mensual, monto_solicitado, tiene_historial_moroso):

    relacion = monto_solicitado / ingreso_mensual

    # Cliente con historial negativo
    if tiene_historial_moroso:

        if relacion > 0.5:
            return (
                "rechazar",
                "relacion monto/ingreso mayor a 0.5 con historial moroso"
            )

        else:
            return (
                "aprobar con condiciones",
                "cliente con historial moroso pero relacion aceptable"
            )

    # Cliente sin historial negativo
    else:

        if relacion <= 0.3:
            return (
                "aprobar",
                "relacion monto/ingreso baja y sin historial moroso"
            )

        elif relacion <= 0.5:
            return (
                "aprobar con condiciones",
                "relacion moderada entre monto solicitado e ingreso"
            )

        else:
            return (
                "rechazar",
                "relacion monto/ingreso demasiado alta"
            )
```

## Simulación y pruebas


```python
# Pruebas del agente de crédito

casos_prueba = [
    # ingreso, monto solicitado, historial moroso
    
    (5000, 1000, False),   # Relación 0.20 -> aprobar
    
    (5000, 2000, False),   # Relación 0.40 -> aprobar con condiciones
    
    (5000, 3000, False),   # Relación 0.60 -> rechazar
    
    (5000, 2500, False),   # Relación 0.50 -> caso límite
    
    (5000, 1000, True),    # Moroso con relación baja
    
    (5000, 4000, True)     # Moroso con relación alta
]


for i, caso in enumerate(casos_prueba, start=1):

    ingreso, monto, moroso = caso

    resultado = agente_credito(
        ingreso,
        monto,
        moroso
    )

    relacion = monto / ingreso

    print(f"Caso {i}")
    print(f"Ingreso mensual: S/. {ingreso}")
    print(f"Monto solicitado: S/. {monto}")
    print(f"Historial moroso: {moroso}")
    print(f"Relación monto/ingreso: {relacion:.2f}")
    print(f"Decisión: {resultado[0]}")
    print(f"Motivo: {resultado[1]}")
    print("-" * 50)
```
